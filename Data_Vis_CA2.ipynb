```python

# ==========================================
# ADVANCED OUTLIER ANALYSIS DASHBOARD
# Full View + Zoomed View
# ==========================================

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# Create 2x2 dashboard
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

fig.patch.set_facecolor("#EEF2F7")

# Custom colour palette
quantity_color = "#3498DB"
price_color = "#9B59B6"

flier_style = {
    "marker": "o",
    "markerfacecolor": "#E74C3C",
    "markeredgecolor": "#922B21",
    "markersize": 4,
    "alpha": 0.65
}

# 1. Full Quantity Distribution
sns.boxplot(
    y=df["Quantity"],
    ax=axes[0, 0],
    color=quantity_color,
    flierprops=flier_style
)

axes[0, 0].set_yscale("symlog", linthresh=10)
axes[0, 0].set_title(
    "Quantity - Full Outlier View",
    fontsize=15, fontweight="bold",
    color="#21618C"
)

# 2. Full Unit Price Distribution
sns.boxplot(
    y=df["UnitPrice"],
    ax=axes[0, 1],
    color=price_color,
    flierprops=flier_style
)

axes[0, 1].set_yscale("symlog", linthresh=1)
axes[0, 1].set_title(
    "Unit Price - Full Outlier View",
    fontsize=15, fontweight="bold",
    color="#76448A"
)

# Calculate zoom ranges (1st to 99th percentile)
q_low, q_high = df["Quantity"].quantile([0.01, 0.99])
p_low, p_high = df["UnitPrice"].quantile([0.01, 0.99])

# 3. Zoomed Quantity Distribution
sns.boxplot(
    y=df["Quantity"],
    ax=axes[1, 0],
    color="#48C9B0",
    flierprops=flier_style
)

axes[1, 0].set_ylim(q_low, q_high)
axes[1, 0].set_title(
    "Quantity - Zoomed Distribution (1%-99%)",
    fontsize=14, fontweight="bold",
    color="#117864"
)

# 4. Zoomed Unit Price Distribution
sns.boxplot(
    y=df["UnitPrice"],
    ax=axes[1, 1],
    color="#F5B041",
    flierprops=flier_style
)

axes[1, 1].set_ylim(p_low, p_high)
axes[1, 1].set_title(
    "Unit Price - Zoomed Distribution (1%-99%)",
    fontsize=14, fontweight="bold",
    color="#9A7D0A"
)

# Professional styling
for ax in axes.flat:
    ax.set_facecolor("white")
    ax.set_ylabel("Value", fontsize=11, fontweight="bold")
    ax.set_xlabel("")
    ax.grid(axis="y", linestyle="--", alpha=0.4)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

# Main heading
fig.suptitle(
    "ONLINE RETAIL DATASET",
    fontsize=22,
    fontweight="bold",
    color="#17202A",
    y=0.99
)

fig.text(
    0.5, 0.955,
    "Advanced Statistical Outlier Analysis",
    ha="center",
    fontsize=13,
    color="#566573"
)

plt.tight_layout(rect=[0, 0, 1, 0.94])
plt.show()
```


### Interpretation of Outlier Analysis

The boxplots above provide a statistical overview of unusual values in the Online Retail dataset. Both Quantity and UnitPrice contain several extreme values outside the normal distribution.

**Quantity Analysis:**
The Quantity boxplot shows unusually high positive and negative values. These may represent bulk purchases, product returns, cancellations or unusual transactions. The zoomed view provides a clearer understanding of the central distribution.

**Unit Price Analysis:**
The UnitPrice boxplot identifies extremely high and negative prices. Some unusual values may be associated with adjustments, fees or exceptional transactions rather than regular product sales.

**Overall Findings:**
The visualisation highlights the presence of outliers in both numerical variables. These records require further investigation before data cleaning. Extreme values should not automatically be removed because some may represent valid business activities.


## 5. Data Cleaning and Preparation

Data cleaning is an essential part of the data analysis process because it helps improve the accuracy and reliability of the results. The initial assessment of the Online Retail dataset identified several issues, including missing values, duplicate records, cancelled transactions and unusual quantities and prices.

In this stage, these issues will be carefully examined and handled to make the dataset more suitable for analysis. Duplicate records will be reviewed, missing values will be assessed, and cancelled transactions and returned items will be checked to understand their impact on sales.

A separate copy of the original dataset will be created so that the original data remains unchanged. The cleaned dataset will then be used for further analysis, visualisations and the development of an interactive dashboard.

```python
# Reload the original dataset

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel("Online Retail.xlsx")

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)
```

```python
# Create a copy of the original dataset
df_clean = df.copy()

# Check the original and copied dataset shapes
print("Original Dataset:", df.shape)
print("Copied Dataset:", df_clean.shape)
```

```python
# Remove duplicate records
df_clean = df_clean.drop_duplicates()

# Check the dataset after removing duplicates
print("Dataset shape after removing duplicates:", df_clean.shape)
print("Remaining duplicate records:", df_clean.duplicated().sum())
```

```python
# Check missing values in the cleaned dataset
missing_values = df_clean.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values)

print("\nTotal Missing Values:", missing_values.sum())
```

```python
# Remove rows with missing product descriptions
df_clean = df_clean.dropna(subset=['Description'])

# Check missing values again
print("Missing values after cleaning:")
print(df_clean.isnull().sum())

print("\nUpdated Dataset Shape:", df_clean.shape)
```

### Handling Missing Values

The dataset initially contained missing values in the Description and CustomerID columns. Rows with missing product descriptions were removed because they could affect product-level analysis. However, missing CustomerID values were retained because these transactions may still contain useful sales information.

After removing missing descriptions, the dataset contained 535,187 rows and 8 columns. CustomerID values will be considered separately during customer-level analysis.

```python
# Identify cancelled transactions
cancelled = df_clean[
    df_clean['InvoiceNo'].astype(str).str.startswith('C')
]

print("Cancelled Transactions:", len(cancelled))
print("\nSample Cancelled Transactions:")
display(cancelled.head())
```

### Review of Cancelled Transactions

The cleaned dataset contains 9,251 cancelled transaction records. These records were identified using invoice numbers starting with the letter 'C'. The sample records also show negative quantities, indicating returns or cancelled orders. These transactions will be kept for now and considered separately to avoid losing useful business information.

```python
# Check negative and zero unit prices
negative_prices = df_clean[df_clean['UnitPrice'] < 0]
zero_prices = df_clean[df_clean['UnitPrice'] == 0]

print("Negative UnitPrice Records:", len(negative_prices))
print("Zero UnitPrice Records:", len(zero_prices))

print("\nNegative Price Records:")
display(negative_prices.head())

print("\nZero Price Records:")
display(zero_prices.head())
```

### Review of Unusual Unit Prices

The analysis identified 2 records with negative unit prices and 1,056 records with zero unit prices. The negative prices were related to bad debt adjustments, while zero-price records may represent free items or special transactions.

These records will be reviewed carefully rather than automatically removed, as they may contain useful business information.
