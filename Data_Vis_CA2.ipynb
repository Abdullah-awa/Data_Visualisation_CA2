### Descriptive Statistical Analysis

Descriptive statistical analysis is used to summarise the main characteristics of a dataset through numerical measurements. It provides an overview of the distribution, central tendency and variation of important variables.

In this analysis, the numerical columns of the cleaned Online Retail dataset will be examined, with particular attention to Quantity, UnitPrice and TotalPrice. These variables are important for understanding customer purchasing activity, product pricing and overall sales performance.

The analysis will examine the mean, median, standard deviation, minimum and maximum values, as well as the quartiles of each numerical variable. These statistical measures will help identify typical transaction values, variations in purchasing quantities and differences in product prices.

The results will provide an initial understanding of the dataset and help guide the selection of appropriate visualisation techniques for the next stages of the assignment.

**Purpose of this analysis:**

- Understand the distribution of numerical variables.
- Compare average and median transaction values.

- Examine the range and spread of sales values.
- Support further sales and product performance analysis.

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the original dataset
df = pd.read_excel("Online Retail.xlsx")

# Create the cleaned dataset
df_clean = df.copy()

# Remove duplicate records
df_clean = df_clean.drop_duplicates()

# Remove rows with missing descriptions
df_clean = df_clean.dropna(subset=['Description'])

# Recreate TotalPrice column
df_clean['TotalPrice'] = (
    df_clean['Quantity'] * df_clean['UnitPrice']
)

print("Dataset loaded successfully!")
print("Cleaned Dataset Shape:", df_clean.shape)
```

```python

# Calculate descriptive statistics for numerical columns

numerical_columns = ['Quantity', 'UnitPrice', 'TotalPrice']

descriptive_stats = df_clean[numerical_columns].describe().T

# Calculate median values
descriptive_stats['Median'] = df_clean[numerical_columns].median()

# Display the results
display(descriptive_stats.round(2))
```

### Interpretation of Descriptive Statistical Analysis

The descriptive statistical analysis summarises three important numerical variables: Quantity, UnitPrice and TotalPrice. The cleaned dataset contains 535,187 records.

**1. Quantity Analysis**

The average quantity per transaction record is 9.67, while the median is 3.00. This difference indicates that the distribution is positively influenced by some unusually large transactions. The standard deviation of 219.06 also shows considerable variation in quantities. Values range from -80,995 to 80,995, indicating the presence of returns, cancellations and extreme transactions.

**2. Unit Price Analysis**

The average unit price is 4.65, compared with a median of 2.08. The standard deviation is 97.36, suggesting substantial variation in product prices. The minimum value is -11,062.06 and the maximum is 38,970.00. These extreme values indicate that some records may relate to adjustments or exceptional transactions rather than ordinary product sales.

**3. Total Price Analysis**

The average transaction value is 18.17, whereas the median is 9.90. The standard deviation is 381.17, showing a wide variation in transaction values. The minimum and maximum values are -168,469.60 and 168,469.60 respectively. Negative transaction values may reflect returns or adjustments, while unusually high values may represent bulk purchases or exceptional transactions.

**Overall Findings**

The analysis shows that all three numerical variables contain considerable variation and extreme values. The mean is higher than the median for each variable, suggesting that unusually large positive values influence the averages. Therefore, the median provides an additional measure of typical transaction values.

These findings highlight the importance of investigating extreme values and separating returns and cancellations from ordinary sales when conducting further business analysis.

```python

# Analyse positive, negative and zero transaction values

positive_transactions = (df_clean['TotalPrice'] > 0).sum()
negative_transactions = (df_clean['TotalPrice'] < 0).sum()
zero_transactions = (df_clean['TotalPrice'] == 0).sum()

transaction_summary = pd.DataFrame({
    'Transaction Type': ['Positive', 'Negative', 'Zero'],
    'Number of Records': [
        positive_transactions,
        negative_transactions,
        zero_transactions
    ]
})

transaction_summary['Percentage'] = (
    transaction_summary['Number of Records']
    / len(df_clean) * 100
)

display(transaction_summary.round(2))
```


### Interpretation of Transaction Value Analysis

The analysis shows that 524,878 records (98.07%) have positive transaction values, representing the majority of recorded sales activity.

A further 9,253 records (1.73%) have negative transaction values. These may represent product returns, cancellations or financial adjustments. The remaining 1,056 records (0.20%) have zero transaction values, which may relate to free items or special transactions.

Overall, positive transactions dominate the dataset. However, negative and zero-value records should be considered carefully when calculating sales performance to avoid misleading business conclusions.


```python
# Calculate total sales revenue for each country

country_sales = (
    df_clean.groupby('Country')['TotalPrice']
    .sum()
    .sort_values(ascending=False)
)

# Display the top 10 countries by sales

top_10_countries = country_sales.head(10)

display(top_10_countries.round(2))
```

```python
# Visualise the top 10 countries by sales revenue

plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_10_countries.values,
    y=top_10_countries.index,
    hue=top_10_countries.index,
    palette='viridis',
    legend=False
)

plt.title('Top 10 Countries by Total Sales Revenue',
          fontsize=16, fontweight='bold')

plt.xlabel('Total Sales Revenue', fontsize=12)
plt.ylabel('Country', fontsize=12)

plt.ticklabel_format(style='plain', axis='x')
plt.grid(axis='x', linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()
```


### Interpretation of Country-wise Sales Analysis

The Netherlands ranks second, followed by EIRE and Germany. The remaining countries contribute comparatively smaller amounts.

**Business Insights:**

```python
# Analyse monthly sales trends

monthly_sales = (
    df_clean
    .groupby(df_clean['InvoiceDate'].dt.to_period('M'))['TotalPrice']
    .sum()
)

# Convert monthly periods into readable dates
monthly_sales.index = monthly_sales.index.to_timestamp()

# Display monthly sales figures
display(monthly_sales.round(2))
```

```python
# Visualise monthly sales trends

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker='o',
    linewidth=2,
    markersize=6,
    color='royalblue'
)

plt.title('Monthly Sales Revenue Trend (2010–2011)',
          fontsize=16, fontweight='bold')

plt.xlabel('Month', fontsize=12)
plt.ylabel('Net Sales Revenue', fontsize=12)

plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
```
