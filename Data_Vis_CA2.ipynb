## 4. Data Quality Assessment

Data quality assessment is an important stage of the analysis because it helps identify problems that may affect the accuracy and reliability of the results.

The original Online Retail dataset was examined to identify missing values, duplicate records, unusual numerical values and potentially invalid transactions.

The initial investigation highlighted missing information in the Description and CustomerID columns. It also revealed negative quantities, negative unit prices and unusually large values that require further investigation.

These issues will be examined before applying appropriate data cleaning techniques. The aim is to improve data quality while preserving valid business information and avoiding unnecessary data loss.

```python
# Check missing values in each column

missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values)

print("\nTotal Missing Values:", df.isnull().sum().sum())
```

```python
# Calculate missing values percentage

missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Percentage (%)': missing_percentage
})

missing_summary
```
